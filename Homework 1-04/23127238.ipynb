## Tiền xử lý

```python
new_cols = {}
for col_name, col_value in cols.items():
    if '/' in col_name:
        new_col_val = []
        for val in col_value:
            try:
                f_val = float(val)
            except:
                f_val = 0
            new_col_val.append(f_val)
        new_cols[col_name] = new_col_val
    elif col_name in ['Started on','Completed']:
        new_col_val = []
        for val in col_value:
            d_value = dt.datetime.strptime(val,'%d/%m/%Y %H:%M')
            new_col_val.append(d_value)
        new_cols[col_name] = new_col_val
    elif col_name == 'Time taken':
        new_col_val = []
        for val in col_value:
            hour = 0;minute=0;secs=0;day=0;
            ls_value = val.split(' ')
            for i in range(len(ls_value)):
                if 'day' in ls_value[i]:
                    day = int(ls_value[i-1])
                elif 'hour' in ls_value[i]:
                    hour = int(ls_value[i-1])
                elif 'min' in ls_value[i]:
                    minute = int(ls_value[i-1])
                elif 'sec' in ls_value[i]:
                    secs = int(ls_value[i-1])
            new_value = (day * 24 + hour) * 60 + minute + secs/60
            new_col_val.append(new_value)
        new_cols[col_name] = new_col_val
    else:
        new_cols[col_name] = col_value
new_cols
```

## Tiền xử lý + phân tích dữ liệu để trả lời cho câu hỏi 1

*Câu hỏi 1: Cột điểm tổng “Grade/10.00” có phân bố các giá trị như thế nào?*

Ở đây không cần tiền xử lý mà sẽ đi luôn vào phân tích dữ liệu.

Đầu tiên, ta sẽ xem thử median của cột "Grade/10.00".

```python
grades = cols['Grade/10.00']
sorted_grades = sorted(grades)
p = 50
i = int(p/100 * (len(sorted_grades)-1))
sorted_grades[i]
```

Để cảm thụ rõ hơn về phân bố các giá trị của cột "Grade/10.00", ta sẽ chia miền giá trị của cột này thành các khoảng (bin) và xem thử mỗi khoảng có bao nhiêu sinh viên.

Giả sử ta sẽ chia đoạn [0, 10] thành 5 bin như sau: [0, 2), [2, 4), [4, 6), [6, 8),[8,10]. Ta có thể lưu kết quả vào một dictionary, trong đó: key là một tuple cho biết 2 đầu của một bin (thay tuple bằng list thì có được không?), value là số lượng sinh viên rơi vào bin này.

***Sinh viên hoàn thành tiếp phần này như bài tập về nhà***

Thay `tuple` bằng `list` thì có được không? Trả lời trực tiếp vào cell này

$\to$ Không được. `tuple` có thể dùng làm key của `dictionary` vì `tuple` là immutable (không thay đổi được), còn `list` là mutable (có thể thay đổi) nên không thể dùng làm key của dictionary.

```python
bins = {
    (0, 2): 0,
    (2, 4): 0,
    (4, 6): 0,
    (6, 8): 0,
    (8, 10): 0
}

for grade in grades:
    if 0 <= grade < 2:
        bins[(0, 2)] += 1
    elif 2 <= grade < 4:
        bins[(2, 4)] += 1
    elif 4 <= grade < 6:
        bins[(4, 6)] += 1
    elif 6 <= grade < 8:
        bins[(6, 8)] += 1
    elif 8 <= grade <= 10:
        bins[(8, 10)] += 1

print("Phân bố điểm theo khoảng:")
for (start, end), count in bins.items():
    percentage = (count / len(grades)) * 100
    if end == 10:
        print(f"[{start}-{end}] điểm: {count} sinh viên ({percentage:.3f}%)")
    else:
        print(f"[{start}-{end}) điểm: {count} sinh viên ({percentage:.3f}%)")
```

## Tiền xử lý + phân tích dữ liệu để trả lời cho câu hỏi 2

*Câu hỏi 2: Câu quiz nào có nhiều bạn làm sai nhất, câu nào có nhiều bạn làm sai nhì, ...?*

Ta có thể lưu kết quả vào một dictionary, trong đó: key là tên của cột ứng với câu quiz, value là điểm trung bình của cột. Dictionary này được sort tăng dần theo value.

***Sinh viên hoàn thành tiếp phần này như bài tập về nhà***

```python
# Câu nào mà có điểm trung bình kiếm được ít nhất thì là câu khó nhất.

question_avg = {}

for col_name, col_vals in cols.items():
    if col_name.startswith('Q.'):
        total = 0
        for val in col_vals:
            total += val
        avg = total / len(col_vals)
        question_avg[col_name] = avg

sorted_questions = sorted(question_avg.items(), key=lambda x: x[1])

print("Thứ tự câu hỏi từ khó nhất đến dễ nhất theo điểm trung bình kiếm được:")
for i, (question, avg_score) in enumerate(sorted_questions, 1):
    print(f"{i}. {question}: {avg_score:.3f} điểm")
```

## Kết luận 1:

**Câu khó nhất (nhiều bạn làm sai nhất - câu có điểm trung bình kiếm được ít nhất): Q. 6 /0.53**

**Câu dễ nhất (ít bạn làm sai nhất - câu có điểm trung bình kiếm được nhiều nhất): Q. 16 /0.53**

---

```python
# Câu nào mà nhiều người bị điểm 0 nhất là sai nhiều nhất.

question_stats = {}

for col_name, col_vals in cols.items():
    if col_name.startswith('Q.'):
        count_zero = 0 
        count_full = 0
        
        for val in col_vals:
            if val == 0:
                count_zero += 1
            elif val == 0.53:
                count_full += 1
        
        total = len(col_vals)
        zero_percent = (count_zero / total) * 100
        full_percent = (count_full / total) * 100
        
        question_stats[col_name] = {
            'zero_count': count_zero,
            'full_count': count_full,
            'total': total,
            'zero_percent': zero_percent,
            'full_percent': full_percent
        }

sorted_questions = sorted(question_stats.items(), key=lambda x: x[1]['zero_count'], reverse=True)

print("Danh sách câu hỏi từ khó nhất đến dễ nhất")
print("(Dựa trên số người làm sai - 0 điểm)")

for i, (question, stats) in enumerate(sorted_questions, 1):
    print(f"{i}. {question}:")
    print(f"Số người sai (0 điểm): {stats['zero_count']}/{stats['total']} ({stats['zero_percent']:.3f}%)")
    print(f"Số người đúng hoàn toàn (5.3 điểm): {stats['full_count']}/{stats['total']} ({stats['full_percent']:.3f}%)")
    print()
```

## Câu hỏi 3
**Có mối tương quan giữa thời gian làm bài và điểm số không?**\
**Mục đích**: Tìm hiểu xem sinh viên làm bài nhanh hay chậm có ảnh hưởng đến điểm số không.

### Đơn giản
#### Bước 1: Tính thống kê cơ bản
- Tính thời gian trung bình của tất cả sinh viên
- Tính thời gian trung vị (median) - đây là mốc chia sinh viên thành 2 nhóm

#### Bước 2: Chia nhóm so sánh
- **Nhóm làm nhanh:** Sinh viên làm bài dưới thời gian trung vị
- **Nhóm làm chậm:** Sinh viên làm bài trên hoặc bằng thời gian trung vị
- Tính điểm trung bình của mỗi nhóm

#### Bước 3: Phân tích kết quả

**Có 3 tình huống có thể xảy ra:**

- **Nhóm làm nhanh có điểm cao hơn:** Có thể sinh viên giỏi nắm kiến thức vững nên làm nhanh và chính xác
- **Nhóm làm chậm có điểm cao hơn:** Có thể sinh viên cẩn thận, suy nghĩ kỹ sẽ đạt kết quả tốt hơn
- **Hai nhóm điểm tương đương:** Không có mối liên hệ rõ ràng giữa thời gian và điểm số

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

### Bước 1: Tính thống kê cơ bản

```python
time_taken = cols['Time taken']
n = len(time_taken)
avg_time = sum(time_taken) / n

sorted_time = sorted(time_taken)
if n % 2 == 0:
    median_time = (sorted_time[n//2 - 1] + sorted_time[n//2]) / 2
else:
    median_time = sorted_time[n//2]

print(f"Thời gian trung bình: {avg_time:.3f} phút")
print(f"Thời gian trung vị: {median_time:.3f} phút")
```

### Bước 2: Chia nhóm so sánh

```python
fast_group = []
slow_group = []

for i in range(len(time_taken)):
    if time_taken[i] < median_time:
        fast_group.append(grades[i])
    else:
        slow_group.append(grades[i])

fast_avg = sum(fast_group) / len(fast_group)
slow_avg = sum(slow_group) / len(slow_group)
```

### Bước 3: Phân tích kết quả

```python
print(f"\nNhóm làm nhanh (< {median_time:.3f} phút):")
print(f"- Số sinh viên: {len(fast_group)}")
print(f"- Điểm trung bình: {fast_avg:.3f}")

print(f"\nNhóm làm chậm (>= {median_time:.3f} phút):")
print(f"- Số sinh viên: {len(slow_group)}")
print(f"- Điểm trung bình: {slow_avg:.3f}")
```

### Nâng cao (không cần thư viện)
- Chia thời gian thành nhiều khoảng (ví dụ: 0-10 phút, 10-20 phút, 20-30 phút...)
- Tính điểm trung bình cho mỗi khoảng thời gian
- Quan sát xu hướng: điểm có tăng hay giảm theo thời gian không?

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

### Bước 1: Chia thời gian thành nhiều khoảng

```python
min_time = min(time_taken)
max_time = max(time_taken)

interval_size = 10
intervals = {}
current = 0
while current < max_time + interval_size:
    intervals[(current, current + interval_size)] = []
    current += interval_size

for i in range(len(time_taken)):
    time = time_taken[i]
    grade = grades[i]
    
    for (start, end) in intervals.keys():
        if start <= time < end:
            intervals[(start, end)].append(grade)
            break
```

### Bước 2: Tính điểm trung bình cho mỗi khoảng

```python
# Tính điểm trung bình cho mỗi khoảng
print("Phân tích theo khoảng thời gian:")
for (start, end), grade_list in sorted(intervals.items()):
    if len(grade_list) > 0:
        avg_grade = sum(grade_list) / len(grade_list)
        print(f"[{start}-{end}) phút: Có {len(grade_list)} sinh viên, điểm trung bình = {avg_grade:.3f}")
    else:
        print(f"[{start}-{end}) phút: Không có sinh viên nào")
```

## Câu hỏi 4
**Có sinh viên nào là điểm ngoại lệ (outliers) không?**\
**Mục đích:** Tìm những sinh viên có điểm số bất thường (quá cao hoặc quá thấp so với mức trung bình).

### Phương pháp 1: Sử dụng khoảng tứ phân vị (IQR - Interquartile Range)

#### Bước 1: Tính các giá trị thống kê
- Sắp xếp tất cả điểm số từ thấp đến cao
- Tính **Q1** (tứ phân vị thứ nhất): Điểm ở vị trí 25% (1/4 đầu dãy số)
- Tính **Q3** (tứ phân vị thứ ba): Điểm ở vị trí 75% (3/4 dãy số)
- Tính **IQR** = Q3 - Q1 (khoảng giữa)

**Ví dụ:** Có 100 sinh viên sau khi sắp xếp
- Q1 = điểm của sinh viên thứ 25 (ví dụ: 4.5)
- Q3 = điểm của sinh viên thứ 75 (ví dụ: 7.5)
- IQR = 7.5 - 4.5 = 3.0

#### Bước 2: Xác định ngưỡng outlier
- **Ngưỡng dưới** = Q1 - 1.5 × IQR
- **Ngưỡng trên** = Q3 + 1.5 × IQR

Tiếp ví dụ trên:
- Ngưỡng dưới = 4.5 - 1.5 × 3.0 = 0
- Ngưỡng trên = 7.5 + 1.5 × 3.0 = 12 (nhưng tối đa là 10)

#### Bước 3: Tìm outliers
- Sinh viên có điểm < ngưỡng dưới → **Outlier thấp** (điểm kém bất thường)
- Sinh viên có điểm > ngưỡng trên → **Outlier cao** (điểm xuất sắc bất thường)

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

### Bước 1: Tính các giá trị thống kê

```python
sorted_grades = sorted(grades)
n = len(sorted_grades)

q1_index = int(0.25 * (n - 1))
Q1 = sorted_grades[q1_index]

q3_index = int(0.75 * (n - 1) + 1)
Q3 = sorted_grades[q3_index]

IQR = Q3 - Q1

print(f"Q1 (tứ phân vị thứ nhất): {Q1}")
print(f"Q3 (tứ phân vị thứ ba): {Q3}")
print(f"IQR (khoảng tứ phân vị): {IQR}")
```

### Bước 2: Xác định ngưỡng outlier

```python
lower_bound = max(0, Q1 - 1.5 * IQR)
upper_bound = min(10, Q3 + 1.5 * IQR)

print(f"\nNgưỡng dưới: {lower_bound:.3f}")
print(f"Ngưỡng trên: {upper_bound:.3f}")
```

### Bước 2: Tìm outliers

```python
low_outliers = []
high_outliers = []

for i, grade in enumerate(grades):
    if grade < lower_bound:
        low_outliers.append((i, grade))
    elif grade > upper_bound:
        high_outliers.append((i, grade))

if not low_outliers and not high_outliers:
    print("\nKhông có outlier theo phương pháp IQR")
else:
    print("\nCó outlier theo phương pháp IQR")
    
    print(f"\nSố sinh viên là outlier thấp: {len(low_outliers)}")
    if low_outliers:
        print("Danh sách (index, điểm):")
        for idx, grade in low_outliers:
            print(f"  Sinh viên thứ {idx+1}: {grade} điểm")

    print(f"\nSố sinh viên là outlier cao: {len(high_outliers)}")
    if high_outliers:
        print("Danh sách (index, điểm):")
        for idx, grade in high_outliers:
            print(f"  Sinh viên thứ {idx+1}: {grade} điểm")
```

### Phương pháp 2: Sử dụng độ lệch chuẩn (Standard Deviation)

#### Bước 1: Tính giá trị trung bình và độ lệch chuẩn
- Tính điểm trung bình (mean)
- Tính độ lệch chuẩn (standard deviation) - đo mức độ phân tán của điểm số

#### Bước 2: Xác định ngưỡng
- **Ngưỡng dưới** = Mean - 2 × Std Dev
- **Ngưỡng trên** = Mean + 2 × Std Dev

**Ví dụ:**
- Điểm trung bình = 6.0
- Độ lệch chuẩn = 1.5
- Ngưỡng dưới = 6.0 - 2 × 1.5 = 3.0
- Ngưỡng trên = 6.0 + 2 × 1.5 = 9.0

#### Bước 3: Tìm outliers
- Điểm < 3.0 hoặc > 9.0 là outliers

***Sinh viên hoàn thành tiếp trong các cell kế như bài tập về nhà***

### Bước 1: Tính giá trị trung bình và độ lệch chuẩn

```python
mean = sum(grades) / len(grades)

variance_sum = 0
for grade in grades:
    variance_sum += (grade - mean) ** 2
variance = variance_sum / (len(grades) - 1)
std_dev = math.sqrt(variance)

print(f"Điểm trung bình (Mean): {mean:.3f}")
print(f"Độ lệch chuẩn (Std Dev): {std_dev:.3f}")
```

### Bước 2: Xác định ngưỡng

```python
lower_bound = mean - 2 * std_dev
upper_bound = mean + 2 * std_dev

print(f"\nNgưỡng dưới: {lower_bound:.3f}")
print(f"Ngưỡng trên: {upper_bound:.3f}")
```

### Bước 3: Tìm outliers

```python
low_outliers = []
high_outliers = []

for i, grade in enumerate(grades):
    if grade < lower_bound:
        low_outliers.append((i, grade))
    elif grade > upper_bound:
        high_outliers.append((i, grade))

if not low_outliers and not high_outliers:
    print("\nKhông có outlier theo phương pháp Std Dev")
else:
    print("\nCó outlier theo phương pháp Std Dev")
    
    print(f"\nSố sinh viên là outlier thấp: {len(low_outliers)}")
    if low_outliers:
        print("Danh sách (index, điểm):")
        for idx, grade in low_outliers:
            print(f"  Sinh viên thứ {idx+1}: {grade} điểm")

    print(f"\nSố sinh viên là outlier cao: {len(high_outliers)}")
    if high_outliers:
        print("Danh sách (index, điểm):")
        for idx, grade in high_outliers:
            print(f"  Sinh viên thứ {idx+1}: {grade} điểm")
```
