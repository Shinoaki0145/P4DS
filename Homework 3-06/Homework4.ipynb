### NumPy Exercises: Student Performance Dataset

#### 📋 Dataset Information

**Dataset**: Students Performance in Exams  
**Source**: https://www.kaggle.com/datasets/spscientist/students-performance-in-exams  
**Size**: 1,000 students  

**Columns**:
- `gender`: Student's gender (male/female)
- `race/ethnicity`: Student's ethnic group (group A, B, C, D, E)
- `parental level of education`: Parent's education level
- `lunch`: Type of lunch (standard/free or reduced)
- `test preparation course`: Completed test prep course (completed/none)
- `math score`: Score in math (0-100)
- `reading score`: Score in reading (0-100)
- `writing score`: Score in writing (0-100)

```python
## NO NEED CODE HERE
import numpy as np

# Load the dataset
data = np.genfromtxt('Data/StudentsPerformance.csv', delimiter=',', 
                     dtype=None, encoding='utf-8', names=True)

def clean_and_convert(arr):
    return np.array([int(str(x).strip('"').strip("'")) for x in arr])

# For easier manipulation, separate numeric scores into a 2D array
# math_scores = data['math_score']
# reading_scores = data['reading_score']
# writing_scores = data['writing_score']

math_scores = clean_and_convert(data['math_score'])
reading_scores = clean_and_convert(data['reading_score'])
writing_scores = clean_and_convert(data['writing_score'])

# Create a 2D array with all scores
scores = np.column_stack([math_scores, reading_scores, writing_scores])

print("Dataset loaded successfully!")
print(f"Number of students: {len(data)}")
print(f"Scores array shape: {scores.shape}")
```

### Task 5: Score Distribution Analysis 
Analyze how scores are distributed across different ranges without using histograms or binning functions

**Instructions**:

1. **Create score bins**: Define 10 ranges for scores:
   - Bin 0: 0-9 points
   - Bin 1: 10-19 points
   - Bin 2: 20-29 points
   - ... and so on ...
   - Bin 9: 90-100 points

2. **Count students in each bin for math scores**: Use a loop to go through all math scores and count how many fall into each bin.

3. **Display the distribution**: Print the count for each bin in a readable format.

4. **Find the most common range**: Determine which bin has the most students (the mode range).

5. **Calculate cumulative distribution**: For each bin, calculate how many students scored at or below that range.

6. **Repeat for reading and writing**: Apply the same analysis to the other two subjects.

7. **Compare distributions**: Identify which subject has the most students in the highest bin (90-100).

```python
#TODO: your code here

def analyze_distribution(subject_scores, subject_name):
# 1. Create an array to store counts for 10 bins
    bins = np.zeros(10, dtype=int)

    # 2. Count students in each bin
    for score in subject_scores:
        bin_idx = min(score // 10, 9)  # 0-9 => bin 0
        bins[bin_idx] += 1
        
    # 3. Display the distribution
    print(f"\n{subject_name} Score Distribution:")
    for i in range(10):
        if i < 9:
            range_str = f"[{i*10}, {i*10+10})"
        else:
            range_str = "[90, 100]"
        print(f"    Bin {i} {range_str}: {bins[i]} students")

    # 4. Find the most common range
    most_common_bin = np.argmax(bins)
    print(f"\nMost common range: Bin {most_common_bin} ({most_common_bin*10}-{most_common_bin*10+9 if most_common_bin < 9 else 100}) with {bins[most_common_bin]} students")

    # 5. Calculate cumulative distribution
    cumulative = np.cumsum(bins)
    print(f"\nCumulative Distribution ({subject_name}):")
    for i in range(10):
        range_str = f"≤{i*10+9}" if i < 9 else "≤100"
        print(f"    {range_str}: {cumulative[i]} students ({cumulative[i]/len(subject_scores)*100:.1f}%)")
        
    return bins

math_bins = analyze_distribution(math_scores, "Math")

# 6. Repeat for reading and writing
reading_bins = analyze_distribution(reading_scores, "Reading")
writing_bins = analyze_distribution(writing_scores, "Writing")

# 7. Compare distributions 
print(f"\nHighest Bin (90-100) Comparison:")
print(f"    Math: {math_bins[9]} students")
print(f"    Reading: {reading_bins[9]} students")
print(f"    Writing: {writing_bins[9]} students")
highest_subject = ["Math", "Reading", "Writing"][np.argmax([math_bins[9], reading_bins[9], writing_bins[9]])]
print(f"    Subject with most students in 90-100 range: {highest_subject}")
```

### Task 6: Student Consistency Analysis
Identify students who have very consistent scores across all three subjects without using variance or standard deviation functions.

**Instructions**:

1. **Calculate score range for each student**: For each student, find the difference between their highest and lowest score among the three subjects.

2. **Define consistency threshold**: Students are "consistent" if their range is 10 points or less (all three scores within 10 points of each other).

3. **Count consistent students**: Count how many students meet the consistency criteria.

4. **Find the most consistent student**: Identify the student with the smallest range (most consistent).

5. **Find the least consistent student**: Identify the student with the largest range (most variable).

6. **Create consistency categories**:
   - Very Consistent: range ≤ 5
   - Consistent: range 6-10
   - Moderate: range 11-20
   - Variable: range 21-30
   - Highly Variable: range > 30

7. **Count students in each category**: Display the distribution.

8. **Analyze patterns**: Among consistent students, which subject combinations tend to be similar?

```python
# TODO: Your code here

# 1. Calculate score range for each student
score_ranges = np.max(scores, axis=1) - np.min(scores, axis=1)

# 2 & 3. Count consistent students 
consistent_mask = score_ranges <= 10
consistent_students = np.sum(consistent_mask)
print(f"Number of consistent students (range ≤ 10): {consistent_students}")

# 4. Find most consistent student
min_range = np.min(score_ranges)
most_consistent_idx = np.argmin(score_ranges)
print(f"\nMost consistent student (index {most_consistent_idx}):")
print(f"    Scores: {scores[most_consistent_idx]}, Range: {min_range}")

# 5. Find least consistent student
max_range = np.max(score_ranges)
least_consistent_idx = np.argmax(score_ranges)
print(f"\nLeast consistent student (index {least_consistent_idx}):")
print(f"    Scores: {scores[least_consistent_idx]}, Range: {max_range}")

# 6. Create consistency categories
categories = {
    "Very Consistent (≤5)": np.sum(score_ranges <= 5),
    "Consistent (6–10)": np.sum((score_ranges >= 6) & (score_ranges <= 10)),
    "Moderate (11–20)": np.sum((score_ranges >= 11) & (score_ranges <= 20)),
    "Variable (21–30)": np.sum((score_ranges >= 21) & (score_ranges <= 30)),
    "Highly Variable (>30)": np.sum(score_ranges > 30)
}

# 7. Count students in each category
print("\nConsistency Distribution:")
for cat, count in categories.items():
    print(f"    {cat}: {count} students")

# 8. Analyze patterns among consistent students
consistent_students = scores[consistent_mask]
print(f"\nAnalyzing {len(consistent_students)} consistent students")

# Check which subject pairs are most similar
math_reading_diff = np.abs(consistent_students[:, 0] - consistent_students[:, 1])
math_writing_diff = np.abs(consistent_students[:, 0] - consistent_students[:, 2])
reading_writing_diff = np.abs(consistent_students[:, 1] - consistent_students[:, 2])

pair_names = ["Math-Reading", "Math-Writing", "Reading-Writing"]
pair_diffs = [math_reading_diff, math_writing_diff, reading_writing_diff]
avg_diffs = [np.mean(d) for d in pair_diffs]

print("Average absolute difference between subjects (consistent students):")
for name, avg in zip(pair_names, avg_diffs):
    print(f"    {name}: {avg:.2f}")

# Find most common similar pair
most_similar_idx = np.argmin(avg_diffs)
print(f"Most similar subject pair: {pair_names[most_similar_idx]} (avg diff: {avg_diffs[most_similar_idx]:.2f})")
```
