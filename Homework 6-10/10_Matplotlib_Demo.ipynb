## Histogram

### Khi nào dùng Histogram?

Histogram dùng để trực quan hóa **phân bố các giá trị** của biến dạng **numerical** (liên tục hoặc rời rạc có nhiều giá trị).

Các loại biến:
- **Numerical** ← Histogram phù hợp
- Categorical ← Bar plot phù hợp

### Nguyên lý hoạt động của Histogram

Histogram hoạt động bằng cách:
1. Phân chia các điểm dữ liệu có giá trị tương tự vào một "bin" chung
2. Đếm số lượng điểm trong mỗi bin
3. **Quan trọng**: Diện tích của mỗi bin tỷ lệ với số lượng điểm dữ liệu (khi dùng `density=True`)

### Đọc dữ liệu điểm quiz từ file và tiền xử lý

````python
# Đọc dữ liệu về kết quả làm quiz của lớp
# (Code lấy từ file 05-Demo.ipynb)

file = open('Data/PythonQuiz_CQ_2025.csv', 'r',encoding='utf-8-sig')

cols = {}
first_line_vals = file.readline().rstrip().split(',')
for first_line_val in first_line_vals:
    cols[first_line_val] = []
for line in file:
    line_vals = line.rstrip().split(',')
    if (line_vals[0] == 'In progress' or line_vals[0] == ''):
        continue
    for i in range(len(line_vals)):
        cols[first_line_vals[i]].append(line_vals[i])

file.close()

cols
````

````python
# Chuyển kiểu dữ liệu của các cột từ str sang kiểu dữ liệu phù hợp
# (Code lấy từ file 04-Demo.ipynb)
new_cols = {}
for col_name, col_vals in cols.items():
    if '/' in col_name:
        new_col_vals = []
        for col_val in col_vals:
            try:
                col_val = float(col_val)
            except:
                col_val = 0
            new_col_vals.append(float(col_val))
        new_cols[col_name] = new_col_vals
    elif col_name in ['Started on', 'Completed']:
        new_col_vals = []
        for col_val in col_vals:
            print(col_val)
            new_col_val = dt.datetime.strptime(col_val,
                                               '%d/%m/%Y %H:%M')
            new_col_vals.append(new_col_val)
        new_cols[col_name] = new_col_vals
    elif col_name == 'Time taken':
        new_col_vals = []
        for col_val in col_vals:
            hour = 0; minute = 0; second = 0
            temp_list = col_val.split(' ')
            for temp_i in range(len(temp_list)):
                if 'hour' in temp_list[temp_i]:
                    hour = int(temp_list[temp_i - 1])
                elif 'min' in temp_list[temp_i]:
                    minute = int(temp_list[temp_i - 1])
                elif 'sec' in temp_list[temp_i]:
                    second = int(temp_list[temp_i - 1])
            new_col_val = hour * 60 + minute + second / 60
            new_col_vals.append(new_col_val)
        new_cols[col_name] = new_col_vals
    else:
        new_cols[col_name] = col_vals
````

````python
cols = new_cols
for col_name, col_vals in cols.items():
    print(f'{col_name} {type(col_vals[0])}')
````

### Tiếp tục với histogram gồm 2 bin [0, 7.5) và [7.5, 10]: phân bố của các giá trị trong bin [7.5, 10]?

Histogram giả định các giá trị được phân bố đều trong mỗi bin. Điều này được thể hiện bởi đường nằm ngang ở trên đầu cột của mỗi bin.

````python
# Thử chi tiết hóa ra bin [7.5, 10]
hist = plt.hist(cols['Grade/10.00'],
                [0, 7.5, 8, 8.5, 9, 9.5, 10],
                edgecolor='black', density=True)
plt.xlabel('Grade')
plt.ylabel('Proportion per grade')
plt.title('Histogram of quiz grades')
plt.xticks(hist[1], rotation=90);
````

Tùy ngữ cảnh cụ thể mà bạn có thể muốn chi tiết hóa ra, có thể không muốn chi tiết hóa ra, nhưng khi nhìn hình của histogram thì bạn nên hiểu về giả định phân bố đều trong mỗi bin của histogram.

### Vẽ histogram của cột điểm "Grade/10.00" với 4 bin [0, 2), [2, 4), [4, 6),  [6, 8), [8, 10] + 3 đường đứng ứng với lower quartile, median, upper quartile

````python
def compute_percentile(data, p):
    sorted_grades = sorted(data)
    i = int(p/100 * (len(sorted_grades)-1))
    return sorted_grades[i]
lower_quartile = compute_percentile(cols['Grade/10.00'], 25)
median = compute_percentile(cols['Grade/10.00'], 50)
upper_quartile = compute_percentile(cols['Grade/10.00'], 75)
lower_quartile, median, upper_quartile
````

````python
hist = plt.hist(cols['Grade/10.00'], [0, 2, 4, 6, 8,10],
                edgecolor='black', density=True)
plt.xlabel('Grade')
plt.ylabel('Proportion per grade')
plt.title('Histogram of quiz grades')
plt.xticks(hist[1])
plt.axvline(lower_quartile,
            color='red', linestyle='solid', linewidth=3,
            label='lower quartile')
plt.axvline(median,
            color='violet', linestyle='dashed', linewidth=3,
            label='median')
plt.axvline(upper_quartile,
            color='yellow', linestyle='dotted', linewidth=3,
            label='upper quartile')
plt.legend();
````

Tại sao diện tích của cột từ upper quartile đến max lại không tương đương với từ median đến upper quartile?

Do giả định phân bố đều của histogram trong bin [5, 10] là không chính xác. Thật ra là có nhiều giá trị tập trung vào vùng từ upper quartile đến max hơn là từ median đến upper quartile.

````python
hist = plt.hist(cols['Grade/10.00'], [0, 2, 4, 6, 8,8.5,9,9.5,10],
                edgecolor='black', density=True)
plt.xlabel('Grade')
plt.ylabel('Proportion per grade')
plt.title('Histogram of quiz grades')
plt.xticks(hist[1])
plt.axvline(lower_quartile,
            color='red', linestyle='solid', linewidth=3,
            label='lower quartile')
plt.axvline(median,
            color='violet', linestyle='dashed', linewidth=3,
            label='median')
plt.axvline(upper_quartile,
            color='yellow', linestyle='dotted', linewidth=3,
            label='upper quartile')
plt.legend();
````

### Đọc dữ liệu đăng ký chuyên ngành K2023 từ file và lấy ra cột các chuyên ngành
Có hai file dành cho hệ CQ và CLC. Cách đọc hơi khác nhau.\
Hai cell tiếp theo sẽ đọc hai file trên.

````python
# Cell sau đọc file Data/CQ2023Majors.txt

file = open('Data/CQ2023Majors.txt', 'r')
majors = []
ignore_lines = []
ignore_lines.append('Không đạt điều kiện xét')
ignore_lines.append('Không đăng ký nguyện vọng chuyên ngành')

major_names = []
major_names.append('Kỹ thuật phần mềm')
major_names.append('Hệ thống thông tin')
major_names.append('Khoa học máy tính + Khoa học dữ liệu')
major_names.append('Công nghệ tri thức + An toàn thông tin')
major_names.append('Công nghệ thông tin')
major_names.append('Thị giác máy tính')
major_names.append('Mạng máy tính và Viễn thông')

for line in file:
    # Dòng ứng với thông tin của một sinh viên là dòng
    # bắt đầu bằng ký tự số
    break_flag = False
    for i_line in ignore_lines:
        if (i_line in line):
            break_flag = True
    if break_flag:
        continue

    if line[0].isdigit():
        # Cắt đuôi "Không đăng ký nguyện vọng chuyên ngành"
        # của dòng để thông tin chuyên ngành sẽ luôn ở cuối dòng

        line = line.rstrip()

        # Để lấy thông tin chuyên ngành thì ta sẽ duyệt từ cuối
        # dòng về đầu và sẽ ngắt khi gặp ký tự viết hoa đầu tiên
        # mà trước đó không có dấu +
        major = ''

        for m in major_names:
            if m in line:
                majors.append(m)
                break
file.close()
majors[0:15]
````

````python
def read_major(file_path):
    file = open(file_path, 'r')
    majors = []

    major_names = []
    major_names.append('Kỹ thuật phần mềm')
    major_names.append('Hệ thống thông tin')
    major_names.append('Khoa học máy tính') #Khoa học máy tính (hướng TTNT)
    major_names.append('Khoa học dữ liệu') #Khoa học dữ liệu
    major_names.append('Công nghệ tri thức')
    major_names.append('Công nghệ thông tin')
    major_names.append('Thị giác máy tính')
    major_names.append('Mạng máy tính và Viễn thông') #Mạng máy tính và Viễn thông (hướng ATTT)

    for line in file:
        # Dòng ứng với thông tin của một sinh viên là dòng
        # bắt đầu bằng ký tự số
        if line[0].isdigit():
            # Cắt đuôi "Không đăng ký nguyện vọng chuyên ngành"
            # của dòng để thông tin chuyên ngành sẽ luôn ở cuối dòng
            line = line.rstrip()

            major = ''
            for m in major_names:
                if m in line:
                    majors.append(m)
                    break
    file.close()
    return majors
````

### Sự khác biệt: Bar Plot vs Histogram

| Bar Plot | Histogram |
|:---------|:----------|
| Trục hoành là biến **categorical** | Trục hoành là biến **numerical** |
| Các bar có thể sắp xếp lại (với nominal) | Các bar có thứ tự cố định |
| Các bar có độ rộng bằng nhau | Các bar có thể có độ rộng khác nhau |
| Có khoảng cách giữa các bar | Các bar nối tiếp nhau |
| Tỷ lệ ∝ **chiều cao** của bar | Tỷ lệ ∝ **diện tích** của bar |

## Ôn lại các câu lệnh đã học

Câu lệnh | Ý nghĩa
:--- | :---
`plt.hist` | Vẽ histogram của biến dạng "numeric"
`plt.bar`, `plt.barh` | Vẽ bar chart của biến dạng "categorical" (cần phải tính số lần hoặc tỉ lệ của các giá trị trước khi vẽ)
`plt.title` | Thêm tựa đề cho hình vẽ
`plt.xlabel` & `plt.ylabel` | Thêm nhãn cho trục x & y
`plt.xlim` & `plt.ylim` | Giới hạn miền giá trị cho trục x & y
`plt.xticks` & `plt.yticks` | Chỉ định các tick cho trục x & y
`plt.axvline` & `plt.axhline` | Vẽ đường thẳng nằm đứng & nằm ngang
`plt.legend` | Thêm chú thích cho ký hiệu, màu, ... trên hình vẽ
`plt.savefig` | Lưu hình vẽ xuống file

## Scatter Plot và Line Plot

Dùng để trực quan hóa **mối quan hệ giữa hai biến liên tục**.

### Scatter Plot
Scatter plots show relationships between two numerical variables as dots, ideal for spotting correlations (positive, negative, none)

````python
x = np.linspace(0, 10, 40)
y = np.sin(x)

plt.figure(figsize=(10, 6))
plt.scatter(x, y, color='blue')
plt.xlabel('X')
plt.ylabel('sin(X)')
plt.title('Basic Scatter Plot')
````

````python
# Format của từng point (size, face color, edge color) có thể được xử lý riêng lẻ
rng = np.random.RandomState(0)
x = rng.randn(100)
y = rng.randn(100)
colors = rng.rand(100)
sizes = 1000 * rng.rand(100)

plt.scatter(x, y, c=colors, s=sizes, alpha=0.3,
            cmap='viridis')
plt.colorbar();  # show color scale
````

### Line Plot

## Box Plot và Violin Plot
### Box Plot
Box plot (biểu đồ hộp) tóm tắt phân bố dữ liệu qua 5 số thống kê:
- Minimum (hoặc Q1 - 1.5*IQR)
- Q1 (25th percentile)
- Median (50th percentile)
- Q3 (75th percentile)
- Maximum (hoặc Q3 + 1.5*IQR)

````python
# Creating dataset
data = cols['Grade/10.00']

# Creating plot
plt.boxplot(data);
````

### Đọc hiểu Box Plot

```
    ┌───────────────┐ ← Maximum (or Q3 + 1.5*IQR)
    │               │
    │               │
────┼───────────────┼──── ← Q3 (75th percentile)
    │               │
    │      ─────    │ ← Median (50th percentile)
    │               │
────┼───────────────┼──── ← Q1 (25th percentile)
    │               │
    └───────────────┘ ← Minimum (or Q1 - 1.5*IQR)
    
    ○               ← Outliers (points beyond whiskers)
```

````python
# Creating dataset
np.random.seed(10)
data = np.random.normal(100, 20, 200)

# Creating plot
plt.boxplot(data);
````

### So sánh nhiều nhóm với Box Plot

````python
# Tạo dữ liệu giả cho nhiều nhóm
data_groups = [
    np.random.normal(100, 10, 200),   # Group A
    np.random.normal(90, 20, 200),    # Group B
    np.random.normal(80, 15, 200),    # Group C
    np.random.normal(110, 12, 200),   # Group D
]

plt.figure(figsize=(10, 6))

bp = plt.boxplot(data_groups, patch_artist=True,
                 tick_labels=['Group A', 'Group B', 'Group C', 'Group D'])

# Tùy chỉnh màu cho mỗi box
colors = ['lightblue', 'lightgreen', 'lightyellow', 'lightpink']
for patch, color in zip(bp['boxes'], colors):
    patch.set_facecolor(color)

plt.ylabel('Value')
plt.title('Comparing Distributions Across Groups');
````

### Violin Plot

Violin plot kết hợp box plot với KDE (Kernel Density Estimation), cho thấy hình dạng phân bố chi tiết hơn.


````python
data = np.random.normal(100, 20, 200)

plt.figure(figsize=(14, 5))

# Box plot
plt.subplot(1, 2, 1)
plt.boxplot(data)
plt.title('Box Plot')
plt.ylabel('Value')

# Violin plot
plt.subplot(1, 2, 2)
plt.violinplot(data, showmedians=True, showmeans=True)
plt.title('Violin Plot')
plt.ylabel('Value')
````

### So sánh Box Plot vs Violin Plot

````python
# Tạo dữ liệu bimodal (hai đỉnh)
np.random.seed(42)
bimodal_data = np.concatenate([
    np.random.normal(30, 5, 500),
    np.random.normal(70, 5, 500)
])

plt.figure(figsize=(15, 5))

# Histogram
plt.subplot(1, 3, 1)
plt.hist(bimodal_data, bins=30, edgecolor='black', alpha=0.7)
plt.title('Histogram - Shows bimodal clearly')

# Box plot
plt.subplot(1, 3, 2)
plt.boxplot(bimodal_data)
plt.title('Box Plot - Hides bimodal nature')

# Violin plot
plt.subplot(1, 3, 3)
plt.violinplot(bimodal_data, showmedians=True)
plt.title('Violin Plot - Shows bimodal clearly')
````

> 💡 **Insight**: Box plot có thể che giấu hình dạng phân bố (như bimodal). Violin plot giúp thấy rõ hơn.

Matplotlib sử dụng cấu trúc phân cấp gồm ba thành phần chính:

```
Figure (Hình vẽ tổng thể)

│   │

│       └── ...
│
└── Các thành phần khác (suptitle, legend tổng,...)
```

### Ba cách sử dụng Matplotlib

#### Cách 1: Pyplot Interface (State-based)

Đây là cách đơn giản nhất, phù hợp cho việc vẽ nhanh. (Trước giờ là làm theo cách này)\
`pyplot`: The State Machine Interface - Giao diện dùng để truy cập figure và axes hiện tại

````python
values = ['a', 'b', 'c']
counts = [1, 2, 3]

plt.bar(values, counts)  # plt tự động tạo figure và axes
plt.title('Bar plot 1')
plt.xlabel('Category')
plt.ylabel('Count');
````

**Ưu điểm**: Đơn giản, nhanh chóng  
**Nhược điểm**: Khó kiểm soát khi vẽ nhiều subplot

#### Cách 2: Object-Oriented Interface
Cách này tường minh hơn và được khuyến khích cho code production.

````python
values = ['a', 'b', 'c']
counts = [1, 2, 3]

fig = plt.figure(figsize=(8, 6))
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])  # [left, bottom, width, height]

ax.bar(values, counts)
ax.set_title('Bar plot 2')
ax.set_xlabel('Category')
ax.set_ylabel('Count')
````

**Ưu điểm**: Kiểm soát tốt, dễ lưu file  
**Nhược điểm**: Code dài hơn

#### More on `figsize`
This creates a new figure (the canvas/window for your plot) with a specific size.

````python
plt.figure(figsize=(10, 6))
#                   ↑    ↑
#                width  height
#               (inches) (inches)
````

| Size | Use Case |
|:-----|:---------|
| `(10, 6)` | Standard wide plot, good for presentations |
| `(8, 8)` | Square, good for scatter plots, heatmaps |
| `(12, 4)` | Very wide, good for time series |
| `(6, 10)` | Tall, good for horizontal bar charts with many categories |

````python
# Create some normally distributed data
mean = [0, 0]
cov = [[1, 1], [1, 2]]
x, y = np.random.multivariate_normal(mean, cov, 3000).T

# Set up the axes with gridspec
fig = plt.figure(figsize=(6, 6))
grid = plt.GridSpec(4, 4, hspace=0.2, wspace=0.2)
main_ax = fig.add_subplot(grid[:-1, 1:])
y_hist = fig.add_subplot(grid[:-1, 0], xticklabels=[], sharey=main_ax)
x_hist = fig.add_subplot(grid[-1, 1:], yticklabels=[], sharex=main_ax)

# scatter points on the main axes
main_ax.plot(x, y, 'go', markersize=3, alpha=0.2)

# histogram on the attached axes
x_hist.hist(x, 40, histtype='stepfilled',
            orientation='vertical', color='blue')
x_hist.invert_yaxis()

y_hist.hist(y, 40, histtype='stepfilled',
            orientation='horizontal', color='red')
y_hist.invert_xaxis()
````
